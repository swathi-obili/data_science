```python
import pandas as pd
df=pd.DataFrame({
    "name":["rahul","joy","kumar","mahi","jhanvi","subham","anit","priya"],
    "age":[25,30,35,40,50,21,43,45],
    "city":["new york","los angles","chicago","houston","hongkong","delhi","mumbai","kolkata"],
    "sal":[50000,60000,70000,80000,30000,32000,45000,87000],
    "experience":[2,5,7,10,15,1,8,20],
    "department":["IT","HR","FINANCE","IT","HR","FINANCE","IT","HR"]
})
print(df)
###here we are going to use .loc function to get both rows and columns together
##df.loc[row_selction,column_selection]
df.loc[1,"sal"]
df.loc[3,"sal"]#here we are accessing specific row and specific column
df.loc[0:3,"name"]#here we are accessing multiple row and specific column
df.loc[0,["name","age","department"]]#here we are accessing  specific row and multiple column
df.loc[df["age"]>25,["name","age","department","sal"]]
mask=df["age"]>25
df.loc[mask,["name","age","department","sal"]]
#df.loc[df["age"]>25 & df["sal"]>50000],["name","age","department","sal"]-it is showing error
### QUERY METHOD LIKE SQL
df.query("age>25 and sal>30000 and city in ['delhi','mumbai']")
con=30 ##whenever we are using variable we need to use @ to access that variable
df.query("age>25 and sal>30000 and city in ['delhi','mumbai']")








print(df[["name","sal"]])
df[df["age"]>40]
print(df.iloc[2:5])
print(df.loc[0:3,["name","departmant"]])
print(df[df["department"]="IT"]["name"])
print(df.iloc[-3:,0:2])
print(df.loc[df["sal"].between(60000,80000),["name","city"]])





```
